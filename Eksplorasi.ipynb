```python
import itertools
from fractions import Fraction
import matplotlib.pyplot as plt
import numpy as np

# Inisialisasi Generator Angka Acak NumPy
rng = np.random.default_rng()

# Ruang sampel dua dadu dibangkitkan dengan itertools.product
ruang = list(itertools.product(range(1, 7), repeat=2))
A = [s for s in ruang if sum(s) == 7]
p_teori = Fraction(len(A), len(ruang))
print("n(S) =", len(ruang), "| n(A) =", len(A))
print("P(jumlah = 7) teori :", p_teori, "=", round(float(p_teori), 4))

# Simulasi Monte Carlo
N = 100_000
lempar = rng.integers(1, 7, size=(N, 2))
jumlah = lempar.sum(axis=1)
p_sim = np.mean(jumlah == 7)
print(
    "P(jumlah = 7) simulasi:",
    round(p_sim, 4),
    "| selisih:",
    round(abs(p_sim - float(p_teori)), 4),
)

# Distribusi seluruh jumlah mata dadu: teori vs simulasi
nilai = np.arange(2, 13)
teori = np.array([sum(1 for s in ruang if sum(s) == k) / 36 for k in nilai])
sim = np.array([np.mean(jumlah == k) for k in nilai])

plt.figure(figsize=(7, 3.8))
plt.bar(nilai - 0.2, teori, width=0.4, label="Teori")
plt.bar(nilai + 0.2, sim, width=0.4, label="Simulasi")
plt.xlabel("Jumlah mata dadu")
plt.ylabel("Probabilitas")
plt.title("Distribusi jumlah dua dadu")
plt.legend()
plt.show()
```

```python
import random
from math import comb, factorial, perm

# Kasus 1: dari 10 server terdapat 3 server rusak. Dipilih acak 4 server.
# P(tepat 2 server rusak) = C(3,2)*C(7,2)/C(10,4)
p_teori = comb(3, 2) * comb(7, 2) / comb(10, 4)
print("P(tepat 2 rusak) teori    :", round(p_teori, 4))

server = ["R"] * 3 + ["B"] * 7
hit = 0
N = 100_000
for _ in range(N):
    if random.sample(server, 4).count("R") == 2:
        hit += 1
print("P(tepat 2 rusak) simulasi :", round(hit / N, 4))

# Kasus 2: PIN 4 digit TANPA angka berulang -> permutasi P(10,4)
print("\nJumlah PIN 4 digit unik :", perm(10, 4))
print("P(menebak PIN dalam 1x) :", round(1 / perm(10, 4), 6))

# Kasus 3: banyak cara menyusun 5 tugas pada 1 prosesor
print("Urutan 5 tugas          :", factorial(5), "cara")
```

```python
# A = bug di modul autentikasi, B = bug di modul pembayaran (independen)
pA, pB = 0.30, 0.40
pAB = pA * pB                       # independen
p_union = pA + pB - pAB             # aturan inklusi-eksklusi
print("P(A n B) =", round(pAB, 4))
print("P(A u B) =", round(p_union, 4))
print("P(tidak ada bug) =", round(1 - p_union, 4))

N = 200_000
a = rng.random(N) < pA
b = rng.random(N) < pB
print("\nSimulasi P(A u B)       :", round(np.mean(a | b), 4))
print("Simulasi P(tidak ada bug):", round(np.mean(~(a | b)), 4))
```

```python
# Data 1000 email: spam/ham dan ada tidaknya tautan mencurigakan
df = pd.DataFrame({
    "Ada link":   [240, 70],
    "Tanpa link": [60, 630],
}, index=["Spam", "Ham"])
total = df.values.sum()
print(df, "\nTotal =", total)

P_spam = df.loc["Spam"].sum() / total
P_link = df["Ada link"].sum() / total
P_spam_link = df.loc["Spam", "Ada link"] / total
P_spam_given_link = P_spam_link / P_link
P_link_given_spam = P_spam_link / P_spam

print("\nP(Spam)               =", round(P_spam, 4))
print("P(Link)               =", round(P_link, 4))
print("P(Spam n Link)        =", round(P_spam_link, 4))
print("P(Spam | Link)        =", round(P_spam_given_link, 4))
print("P(Link | Spam)        =", round(P_link_given_spam, 4))

# Uji independensi: independen jika P(A n B) = P(A) * P(B)
print("\nP(Spam)*P(Link)       =", round(P_spam * P_link, 4))
print("Independen?           =", np.isclose(P_spam_link, P_spam * P_link))
```

```python
# Tiga server melayani beban 50%, 30%, 20% dengan laju error 1%, 2%, 5%
nama = ["Server A", "Server B", "Server C"]
prior = np.array([0.50, 0.30, 0.20])
likelihood = np.array([0.01, 0.02, 0.05])

p_error = np.dot(prior, likelihood)             # hukum probabilitas total
posterior = prior * likelihood / p_error        # Bayes
print("P(error) =", round(p_error, 4))
print(pd.DataFrame({"Prior": prior, "P(error|server)": likelihood,
                    "Posterior P(server|error)": posterior.round(4)}, index=nama))

# Verifikasi dengan simulasi
N = 500_000
s = rng.choice(3, size=N, p=prior)
err = rng.random(N) < likelihood[s]
print("\nSimulasi P(error)         :", round(err.mean(), 4))
print("Simulasi P(Server C|error):", round(np.mean(s[err] == 2), 4))

fig, ax = plt.subplots(1, 2, figsize=(9, 3.6))
ax[0].bar(nama, prior, color="steelblue"); ax[0].set_title("Prior (beban)")
ax[1].bar(nama, posterior, color="indianred"); ax[1].set_title("Posterior (setelah ada error)")
plt.tight_layout(); plt.show()
```

```python
prevalensi, sensitivitas, fpr = 0.01, 0.95, 0.05

def posterior_virus(prev, sens=sensitivitas, fp=fpr):
    return prev * sens / (prev * sens + (1 - prev) * fp)

print("P(virus | alarm) =", round(posterior_virus(prevalensi), 4))

# Simulasi 1.000.000 file
N = 1_000_000
virus = rng.random(N) < prevalensi
alarm = np.where(virus, rng.random(N) < sensitivitas, rng.random(N) < fpr)
print("Simulasi P(virus | alarm) =", round(virus[alarm].mean(), 4))

# Pengaruh prevalensi terhadap posterior
prev = np.linspace(0.001, 0.5, 200)
plt.figure(figsize=(7, 3.8))
plt.plot(prev, posterior_virus(prev))
plt.scatter([prevalensi], [posterior_virus(prevalensi)], color="red", zorder=3, label="Prevalensi 1%")
plt.xlabel("Prevalensi virus P(virus)"); plt.ylabel("P(virus | alarm)")
plt.title("Alarm positif belum tentu berarti terinfeksi"); plt.legend(); plt.show()
```

```python
def monty_hall(ganti, n=100_000):
    menang = 0
    for _ in range(n):
        hadiah = rng.integers(3)
        pilihan = rng.integers(3)
        buka = next(d for d in range(3) if d != pilihan and d != hadiah)
        if ganti:
            pilihan = next(d for d in range(3) if d != pilihan and d != buka)
        menang += (pilihan == hadiah)
    return menang / n

print("P(menang | tetap) simulasi:", round(monty_hall(False), 4), "| teori:", round(1/3, 4))
print("P(menang | ganti) simulasi:", round(monty_hall(True), 4), "| teori:", round(2/3, 4))
```

```python
# 20 paket dikirim, tiap paket hilang dengan peluang 0,10
n, p = 20, 0.10
X = stats.binom(n, p)
print("P(X = 2)  =", round(X.pmf(2), 4))
print("P(X <= 2) =", round(X.cdf(2), 4))
print("P(X >= 5) =", round(X.sf(4), 4))
print("E(X) = np     =", X.mean(), "| Var(X) = npq =", round(X.var(), 4))

data = X.rvs(100_000, random_state=1)
print("Rata-rata sampel =", round(data.mean(), 4), "| Var sampel =", round(data.var(), 4))

k = np.arange(0, 11)
plt.figure(figsize=(7, 3.8))
plt.bar(k, X.pmf(k), alpha=0.6, label="PMF teori")
plt.plot(k, [np.mean(data == i) for i in k], "ro-", label="Simulasi")
plt.xlabel("Jumlah paket hilang"); plt.ylabel("P(X = k)")
plt.title("Binomial(n=20, p=0,10)"); plt.legend(); plt.show()
```

```python
# Rata-rata 3 permintaan masuk per menit
lam = 3
Y = stats.poisson(lam)
print("P(Y = 5)  =", round(Y.pmf(5), 4))
print("P(Y >= 8) =", round(Y.sf(7), 4))
print("E(Y) =", Y.mean(), "| Var(Y) =", Y.var())

# Hampiran: Binomial(n=1000, p=0,003) ~ Poisson(lambda = np = 3)
B = stats.binom(1000, 0.003)
k = np.arange(0, 13)
print("\nSelisih maksimum |PMF Binomial - PMF Poisson| =", round(np.max(np.abs(B.pmf(k) - Y.pmf(k))), 6))

plt.figure(figsize=(7, 3.8))
plt.bar(k - 0.2, Y.pmf(k), width=0.4, label="Poisson(3)")
plt.bar(k + 0.2, B.pmf(k), width=0.4, label="Binomial(1000; 0,003)")
plt.xlabel("k"); plt.ylabel("Probabilitas"); plt.title("Hampiran Poisson"); plt.legend(); plt.show()
```

```python
# Peluang sukses tiap percobaan kirim = 0,7
p = 0.7
G = stats.geom(p)                    # X = percobaan ke-berapa sukses pertama (mulai dari 1)
print("P(sukses pada percobaan ke-3) =", round(G.pmf(3), 4))
print("P(butuh <= 2 percobaan)       =", round(G.cdf(2), 4))
print("E(X) = 1/p =", round(G.mean(), 4), "| Var = (1-p)/p^2 =", round(G.var(), 4))

data = G.rvs(100_000, random_state=2)
print("Rata-rata percobaan (simulasi) =", round(data.mean(), 4))

k = np.arange(1, 9)
plt.figure(figsize=(7, 3.8))
plt.bar(k, G.pmf(k)); plt.xlabel("Percobaan ke-k"); plt.ylabel("P(X = k)")
plt.title("Geometrik(p = 0,7)"); plt.show()
```

```python
# Batch 50 chip berisi 5 cacat. Diambil 6 chip tanpa pengembalian.
M, n_cacat, N_ambil = 50, 5, 6
H = stats.hypergeom(M, n_cacat, N_ambil)
print("P(tidak ada cacat) =", round(H.pmf(0), 4))
print("P(minimal 1 cacat) =", round(H.sf(0), 4))
print("E(X) =", round(H.mean(), 4), "| Var(X) =", round(H.var(), 4))

# Pembanding: jika dianggap dengan pengembalian -> Binomial(6, 5/50)
B = stats.binom(N_ambil, n_cacat / M)
print("\nP(minimal 1 cacat) bila dianggap Binomial =", round(B.sf(0), 4))

k = np.arange(0, 6)
plt.figure(figsize=(7, 3.8))
plt.bar(k - 0.2, H.pmf(k), width=0.4, label="Hipergeometrik")
plt.bar(k + 0.2, B.pmf(k), width=0.4, label="Binomial")
plt.xlabel("Jumlah chip cacat"); plt.ylabel("Probabilitas"); plt.legend()
plt.title("Hipergeometrik vs Binomial"); plt.show()
```

```python
# Jumlah bug kritis per rilis (distribusi diskrit buatan)
x = np.array([0, 1, 2, 3, 4])
p = np.array([0.40, 0.30, 0.15, 0.10, 0.05])
assert np.isclose(p.sum(), 1)

E = np.sum(x * p)
E2 = np.sum(x**2 * p)
Var = E2 - E**2
print("E(X)   =", round(E, 4))
print("E(X^2) =", round(E2, 4))
print("Var(X) =", round(Var, 4), "| Simpangan baku =", round(np.sqrt(Var), 4))

# Cek memakai scipy.stats.rv_discrete
D = stats.rv_discrete(values=(x, p))
print("scipy: mean =", round(D.mean(), 4), "| var =", round(D.var(), 4))

# Biaya perbaikan Rp2 juta per bug -> E(biaya) = 2*E(X)   (sifat linearitas)
print("Harapan biaya perbaikan per rilis = Rp", round(2 * E, 2), "juta")

sim = D.rvs(size=100_000, random_state=3)
print("Rata-rata simulasi =", round(sim.mean(), 4))

plt.figure(figsize=(7, 3.8))
plt.stem(x, p); plt.xlabel("Jumlah bug kritis"); plt.ylabel("P(X = x)")
plt.title("Distribusi jumlah bug kritis"); plt.show()
```

```python
# Waktu hingga request ke-3, tiap antar-kedatangan ~ Exp(mean 2 detik)
# => Gamma(shape=3, scale=2)
Gm = stats.gamma(a=3, scale=2)
print("P(X <= 5) =", round(Gm.cdf(5), 4))
print("E(X) = a*scale =", Gm.mean(), "| Var = a*scale^2 =", Gm.var())

# Verifikasi: jumlah 3 variabel Eksponensial
jumlah3 = stats.expon(scale=2).rvs((100_000, 3), random_state=7).sum(axis=1)
print("Simulasi jumlah 3 Eksponensial: mean =", round(jumlah3.mean(), 4), "| P(X<=5) =", round(np.mean(jumlah3 <= 5), 4))

xs = np.linspace(0, 25, 400)
plt.figure(figsize=(7, 3.8))
plt.hist(jumlah3, bins=60, density=True, alpha=0.5, label="Jumlah 3 Eksponensial")
plt.plot(xs, Gm.pdf(xs), "r-", label="Gamma(3, scale=2)")
plt.legend(); plt.xlabel("Detik"); plt.title("Gamma sebagai jumlah Eksponensial"); plt.show()
```

```python
# (1) Sifat distribusi: Chi-square dengan df = 3
k = 3
C2 = stats.chi2(df=k)
print("E(X) = k =", C2.mean(), "| Var(X) = 2k =", C2.var())
print("Nilai kritis (persentil ke-95) =", round(C2.ppf(0.95), 4))
print("P(X > 7,815) =", round(C2.sf(7.815), 4))

# Chi-square = jumlah kuadrat k variabel normal baku -> cek dengan simulasi
z = rng.standard_normal((100_000, k))
q = (z**2).sum(axis=1)
print("Simulasi: mean =", round(q.mean(), 4), "| P(X > 7,815) =", round(np.mean(q > 7.815), 4))

# (2) Uji kecocokan: apakah dadu adil? 120 lemparan
observasi = np.array([18, 22, 16, 25, 19, 20])
harapan = np.full(6, observasi.sum() / 6)
chi_hitung = np.sum((observasi - harapan) ** 2 / harapan)
hasil = stats.chisquare(observasi, harapan)
print("\nChi-square hitung (manual) =", round(chi_hitung, 4))
print("scipy: statistik =", round(hasil.statistic, 4), "| p-value =", round(hasil.pvalue, 4))
print("Nilai kritis (df=5, alpha=0,05) =", round(stats.chi2.ppf(0.95, 5), 4))

xs = np.linspace(0, 15, 400)
plt.figure(figsize=(7, 3.8))
plt.hist(q, bins=60, density=True, range=(0, 15), alpha=0.5, label="Simulasi")
plt.plot(xs, C2.pdf(xs), "r-", label="PDF Chi-square (df=3)")
plt.axvline(C2.ppf(0.95), color="blue", ls="--", label="Nilai kritis 95%")
plt.xlabel("x"); plt.legend(); plt.title("Distribusi Chi-square"); plt.show()
```
